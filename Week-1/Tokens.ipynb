# Tokens

## Why Not Just Use Characters or Words?

### Attempt 1: Character-by-character
- Small vocabulary (~100 possible characters: letters, punctuation, etc.) → simple, memory-efficient.
- **Problem**: too much burden on the model — it had to learn both *how to spell words* and *what those words mean*, all from individual letters. Too hard to train well.

### Attempt 2: Whole words
- More intuitive — one input = one word.
- **Problem**: vocabulary explodes — huge number of possible words (plus names, places, rare proper nouns).
- Rare words had to be left out of the vocabulary → model couldn't understand them.

### The Breakthrough: Tokens
- A **middle ground** between characters and words.
- A token can be:
  - A whole word
  - A fragment/piece of a word
  - Occasionally, two common words combined
- This keeps the vocabulary **limited/manageable**, while still letting the model recognize meaningful word pieces (like stems, prefixes, suffixes).

> 💡 There's nothing "fundamental" about tokens — it's an efficiency compromise, similar to how the Transformer architecture is an efficiency choice, not a theoretical necessity.

---

## Why Tokens Work Well
- Efficient — smaller vocabulary than whole-word approaches.
- Learns concepts quickly.
- Naturally captures **word stems + modifiers** — e.g., a base chunk of characters followed by a chunk that modifies its meaning (which fits how language/spelling naturally works).

## Tokens as Model Input
- Tokens are the **very first input** into an LLM (or any deep neural network).
- Technically, what's passed in is the **token ID** — a number representing which token it is (based on a fixed vocabulary).
- **Tokens ≠ Vectors**:
  - Token ID = the raw input identifier (comes first).
  - Vectors = come later in the network (a separate, deeper concept — covered later).

Text → [Tokenizer splits text into chunks] → Token IDs (numbers) → fed into the neural network

## Key Terms
- **Token**: a chunk of text (word, sub-word fragment, or occasionally multiple words) used as the basic unit of input/output for an LLM.
- **Token ID**: the numeric identifier representing a specific token in the model's vocabulary.
- **Vector**: a different, later-stage representation in the network (not the same as a token ID).
- **Tokenizer**: the tool/algorithm that converts raw text into tokens.
