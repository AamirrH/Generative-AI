# LSTMs, Emergent Intelligence & Agentic AI

## Before the Transformer: LSTM
- **LSTM (Long Short-Term Memory)**: a type of recurrent neural network (RNN), the dominant model before Transformers.
- Arguably had **deeper understanding** of sequence relationships than Transformers.
- **Problem**: had to process step-by-step (each output fed into the next) → **couldn't be parallelized** → slow and hard to train at scale.
- **Transformer = a simplification** — gave up some of that "cleverness" but gained **massive parallelization**, which mattered far more in practice.
- This is the meaning behind the paper title *"Attention Is All You Need"* — you don't need all the complex machinery of LSTMs; simple attention, scaled up, is enough.

## World Reaction (2023)
- Initial shock/hype — "transformer" became a mainstream buzzword.
- Backlash followed — notably the **"Stochastic Parrots"** paper, warning that people were mistaking statistical next-word prediction for real understanding/truth.
- That concern hasn't aged well, given how capable models have become.

## The Real Mystery: Emergent Intelligence
- It's **not surprising** that a model can predict plausible-sounding next words (that's just good predictive text).
- What **is surprising**: the predicted text so often turns out to be **actually correct** (e.g., solving real math problems), not just plausible-sounding.
- We understand the *mechanics* (statistics, token prediction) but not fully *why* this accuracy emerges.
- This phenomenon — intelligence-like behavior emerging from scale — is called **emergent intelligence**.

## Key Trend Timeline

### 1. Prompt Engineering
- Was briefly a dedicated high-paying job.
- Now a basic skill everyone has — no longer a specialized role.

### 2. Copilots
- Human + LLM collaboration tools (Microsoft Copilot, GitHub Copilot) — AI assisting alongside humans on documents/code.

### 3. Context Engineering (the "new prompt engineering")
- About giving the LLM **all the right information** to succeed — not just clever phrasing.
- Includes: relevant business/domain data in the prompt, and equipping the model with **tools** it can use.
- Core idea: the better/more relevant the input sequence, the better the output. 
  - E.g., if answering ticket price questions, the actual prices need to be in the input for the model to answer correctly.

### 4. Agentic AI (current hot topic)
Two common definitions:
1. **LLM controls the workflow** — decides what happens next, including calling other LLMs or tools.
2. **LLM in a loop with tools** — repeatedly calls itself, plans, acts, checks results, and continues (e.g., Claude Code's to-do list style execution).

- **Autonomy**: the LLM's output can include instructions on what to do *next*, effectively letting it choose its own next steps — this is what makes a system "agentic."
- Despite sounding complex, it's still fundamentally: **input sequence → output sequence** — just that the output can trigger further actions.

---

## Key Terms
- **LSTM**: older recurrent network architecture, powerful but hard to parallelize.
- **Emergent intelligence**: unexplained tendency for large-scale models to produce *accurate*, not just plausible, outputs.
- **Context engineering**: structuring the best possible input (data + tools) for an LLM to succeed at a task.
- **Agentic AI**: an LLM that operates in a loop, uses tools, and decides its own next actions.
- **Autonomy**: the model choosing what to do next based on its own output.
