## What Frontier Models Are Great At
- **Synthesizing information** — summarizing, answering detailed questions, weighing pros/cons with well-structured answers.
- **Content generation** — drafting emails, presentations, project plans; great for brainstorming/starting new work.
- **Coding** — writing, testing, and iterating on code in loops; has largely replaced Stack Overflow as the go-to debugging resource.

---

## Where They Fall Short

### 1. Knowledge Gaps
- Superhuman in some domains, weak in others — knowledge isn't uniform.
- **Knowledge cutoff**: models only know information up to their training date.
  - Example: a model may confidently insist a newer model name "doesn't exist" and suggest an outdated one instead.
- Features like **web search** in products (ChatGPT, etc.) are *not* part of the model itself — they're extra engineering built around it.

### 2. Hallucination
- Models only predict the **most plausible next word/token** — they aren't verifying truth.
- They're trained to sound **confident**, whether right or wrong.
- It's actually surprising they're *as accurate* as they are, given they're just predicting plausible text.
- When wrong, they're wrong **confidently and convincingly** — which is dangerous.

### 3. Best for Seniors, Riskier for Juniors
- Contrary to early expectations, LLMs help **experienced developers** most — they can spot and correct mistakes.
- **Junior/inexperienced users** are more at risk of being misled, since they can't tell when the model has gone off track.

#### Example (from the course)
- A student passed a **base model** (not chat/instruct variant) chat-style prompts (system + user prompt).
- It failed since base models aren't trained to handle that structure.
- Instead of identifying the real issue (wrong model type used), an LLM helping the student wrote pages of complex, unnecessary code trying to *simulate* chat behavior on top of the base model.
- The student, not knowing better, assumed progress was being made — the real root cause (simple mistake: wrong model name/type) was never addressed.
- **Lesson**: LLMs tend to patch surface symptoms rather than step back and question the actual root cause — especially in coding tasks.

---

## Key Takeaway
> LLMs are like a **tireless junior analyst** — extremely capable and hardworking, but need **supervision**. It's your job to check their work, catch when they go off-track, and make sure they're not jumping to conclusions.
