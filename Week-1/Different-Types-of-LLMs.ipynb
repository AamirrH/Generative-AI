# Types of LLMs: Base, Chat, and Reasoning Models

There are **three breeds of LLMs**, based on what they've been trained to do.

---

## 1. Base Models
- Takes a sequence of text as input and **predicts what comes next** — that's it.
- No concept of "chatting" or "answering questions" — just sequence completion.
- **Everyday example**: predictive text on your phone keyboard. You type "Hello there", tap the suggested next word repeatedly — that's a base model in action.
- Before ChatGPT (2022), models like **GPT-3** were base models.
- **Trick people used to get Q&A behavior out of base models**:

Q: example question\
A: example answer\
Q: example question\
A: example answer\
Q: your actual question\
A:

This "primed" the model into a question-answering pattern purely through sequence prediction.
- **When to use a base model**: when you want to fine-tune/train a model on a completely new skill or task construct — you don't want it biased toward chat or reasoning behavior already baked in.

---

## 2. Chat Models (Instruct Models)
- OpenAI found that training a model on structured **message → response → message → response** data made it much better for conversation.
- This introduced the now-standard structure:
  - **System prompt** — overall instructions/context for the whole conversation.
  - **User prompt** — the user's message.
  - **Assistant reply** — the model's response.
- Training technique used: **RLHF (Reinforcement Learning from Human Feedback)**.
- This is what turned base GPT into **ChatGPT**.
- Also called **Instruct models**.

### Chain of Thought Prompting
- A prompt engineering trick: simply add **"please think step by step"** to your prompt.
- Improves output quality because the model generates a more methodical sequence.
- This observation inspired training models to *natively* think step-by-step → leading to **reasoning models**.

---

## 3. Reasoning Models (Thinking Models)
- Trained to **output their thinking/reasoning steps first**, then give a final answer.
- Essentially, chain-of-thought behavior baked into training rather than prompted manually.

### Reasoning Budget / Reasoning Effort
- The amount of "thinking" a reasoning model does before answering.
- Can be controlled — more reasoning = generally better performance on hard problems, but slower and costlier.

### Budget Forcing (S1 Paper, Jan 2025)
- A surprisingly simple technique to make a model reason more.
- Discovery: inserting the word **"wait"** into the model's thinking trace causes it to pause, reconsider, and re-evaluate its reasoning — improving output quality.
- Example: model reasoning says "I should consider X..." → insert "wait" → model continues "wait, I should reconsider... let me re-check..."
- Counterintuitively simple, but effective. (Paper: **S1**)

---

## Hybrid Models
- The newest category — models that **decide how much to reason** based on the query.
- Simple/chat-style questions → little to no reasoning (acts like a chat model).
- Complex/puzzle-like questions → engages deeper reasoning.
- Examples: **Gemini 2.5 Pro**, **GPT-5**.
- Some open source models offer separate "hybrid" and "chat-only" variants, so you can force chat-only mode if needed.

---

## Comparing the Three Types

| Model Type | Strengths | Weaknesses |
|---|---|---|
| **Base** | Best starting point for training new skills/tasks | Not usable directly for chat/Q&A |
| **Chat** | Faster, cheaper (no reasoning trace), better for interactive use, often more fluid/creative for content generation (emails, writing) | Weaker at complex problem-solving/puzzles |
| **Reasoning** | Better at problem-solving, puzzles, scores higher on intelligence benchmarks | Slower, more expensive (extra "thinking tokens"), can feel "overthought" or cold for creative writing |

> 📌 **Note**: The idea that chat models are better for creative writing is **anecdotal**, not backed by strong metrics — worth testing yourself.

---

## Key Terms
- **Base model**: raw next-token predictor, no chat/reasoning structure.
- **Chat/Instruct model**: trained on structured conversation (system/user/assistant) via RLHF.
- **Reasoning model**: trained to think step-by-step before answering.
- **Hybrid model**: dynamically decides how much reasoning to apply.
- **Reasoning budget / effort**: how much "thinking" a model does.
- **Budget forcing**: technique (e.g., inserting "wait") to force a model to reason more.
