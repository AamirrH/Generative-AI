# Parameters & Scaling (Training-Time vs Inference-Time)

## What Are Parameters?
- The internal numeric values a neural network learns during training — they determine how it processes input and generates output.
- Traditional ML models: ~20–200 parameters (e.g., factors for predicting creditworthiness).
- Deep learning changed the scale entirely.

## Growth of Parameters Over Time

| Model | Parameters |
|---|---|
| GPT-1 | 117 million |
| GPT-2 | 1.5 billion |
| GPT-3 | 175 billion |
| GPT-4 | ~1.76 trillion |
| Latest frontier models | Unknown (likely tens of trillions) — not disclosed |

- General rule: **more parameters → more capacity to absorb training data → more "intelligent."**
- However, recent models show we can do **more with fewer parameters** through efficiency gains:
  - e.g., Gemma's smallest version (270M params) is **smaller than GPT-2** but **far more capable**.

## Model Size Tiers
Many providers release the same model family in multiple sizes (different parameter counts):
- **GPT-5**: Nano → Mini → Full
- **Claude**: Haiku (small) → Sonnet (mid) → Opus (large)
- Bigger size = more parameters = higher compute cost = higher price.

## Two Ways to Improve a Model's Performance

### 1. Training-Time Scaling
- Making the model bigger (more parameters) and training it on more data.
- Historically the main way labs improved models ("bigger is better").
- Related concept: **Chinchilla scaling laws** — parameter count should roughly match how much training data the model can effectively absorb.

### 2. Inference-Time Scaling
- Improving performance **at the point of use** (inference = running the trained model), without changing the model itself.
- Techniques include:
  - **Reasoning**: prompting the model to "think step by step" before answering (covered in reasoning models).
  - **Adding more context to the input**: e.g., feeding in relevant data like ticket prices so outputs are more accurate — this is the foundation of **RAG (Retrieval-Augmented Generation)**.

Training-Time Scaling: Bigger model + more training data → smarter model (before deployment)
Inference-Time Scaling: Better prompting/context at runtime → smarter answers (after deployment)

- These are **two independent ("orthogonal") levers** — you can use both together.
- Historically, focus was almost entirely on training-time scaling. In the last ~1–2 years, **inference-time scaling** (RAG, then reasoning techniques) has become just as important.

## Key Terms
- **Parameter**: a learned numeric value inside a neural network that shapes its behavior.
- **Training-time scaling**: improving a model by making it bigger / training on more data.
- **Inference-time scaling**: improving output quality at usage time (e.g., reasoning, adding context) without changing the model itself.
- **Mixture of Experts (MoE)**: architecture where a large model contains many smaller specialized sub-models, only some active per query.
- **Chinchilla scaling laws**: guideline relating model size to the ideal amount of training data.
