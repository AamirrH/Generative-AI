```python
import tiktoken
```

```python
## Select the type of encoding and the encoder model
encoding = tiktoken.encoding_for_model("gpt-5")

text = "My Name is Aamir Hussain"

## Encode The Text

token = encoding.encode(text)

print(token)
```

```python
## [TokenId1,TokenId2...]

for token_id in token:
    decoded_text = encoding.decode([token_id])
    print(token_id,":",decoded_text)
```

```python
## Decoding takes a list of token as an argument and returns a string.
encoding.decode([5444, 7317, 382, 355, 114042, 171732, 524])
```
