# LLM Engineering Basics


## Open Source LLMs & Ways to Use Models

## Open Source vs Closed Source
- **Closed source (frontier) models**: GPT, Claude, Gemini — proprietary, accessed only via API/product.
- **Open source models**: weights are publicly released, can be downloaded and run yourself.

---

## Major Open Source Model Families

### 1. Meta — Llama
- Meta was the **first major company** to open source frontier-level models.
- **Llama 3.2**: comes in small sizes (1B / 3B parameters) — easy to run locally. Sometimes called an **SLM (Small Language Model)**, though 3B is still fairly large.
- **Llama 4**: Meta's larger, more powerful open source model.

### 2. Mistral (France)
- Known for **Mixture of Experts (MoE)** architecture — model contains multiple smaller "expert" sub-models, and routes each query to the relevant expert(s) instead of using the whole network every time.

### 3. Qwen (Alibaba Cloud, China)
- Very powerful, somewhat underrated compared to Llama.
- Worth trying — strong performance despite less hype.

### 4. Gemma (Google)
- Open source "cousin" of Gemini.
- Comes in many sizes, including a tiny **270M parameter** version (very small, limited but functional).

### 5. Phi (Microsoft)
- Phi-4 is the latest at time of recording.
- Strong at **tool calling** and commercial/enterprise use cases.

### 6. DeepSeek (DeepSeek AI)
- Caused major excitement ~6 months before this video due to **training efficiency**, not raw power.
  - OpenAI reportedly spent **$100M+** training GPT-level models.
  - DeepSeek reportedly spent **~$4M** to reach near-frontier performance.
- Main model: **671B parameters** — too large for local/consumer use.
- Smaller "DeepSeek" variants you can run locally are **not actually DeepSeek architecture**:
  - They are **Llama** or **Qwen** models, further trained on **synthetic data generated by the big DeepSeek model**.
  - This technique is called **Distillation**: using a large "teacher" model's outputs to train a smaller "student" model.

### 7. GPT-OSS (OpenAI)
- OpenAI's open source release, came out a few weeks before this recording.
- Likely influenced by competitive pressure from DeepSeek's open release.
- Comes in two sizes:
  - Smaller version (~20B) — runnable on consumer hardware.
  - Larger version (120B) — too big for most personal computers.

## Three Ways to Use LLMs

### 1. Packaged Products
- User-facing applications built **on top of** a model (e.g., ChatGPT, Claude.ai).
- Includes extra engineered features: memory, web search, UI, etc.
- You're using a **product**, not directly using the raw model.

### 2. Cloud APIs
- Direct API calls to a hosted model (e.g., calling OpenAI's or Anthropic's API).
- Can also go through **frameworks** that manage these API calls for you.
- **Managed cloud services** also fall here — e.g., Amazon Bedrock, Google Vertex AI, Azure ML — these host and serve models for you.
- **Groq** (spelled with a "q", different from Elon Musk's Grok with a "k"): a service offering very fast inference of open source models using specialized proprietary hardware.
- **OpenRouter**: a platform that routes your API requests to many different model providers.

### 3. Direct Inference (Running Models Yourself)
Only possible with **open source models**. Two common tools:

#### Hugging Face Transformers Library
- Directly run the actual model code (Python/C++) and weights yourself in a Python environment.
- Most "raw"/flexible option — you're running the real underlying code.

#### Ollama
- A **packaged product** for easily running open source models locally.
- Model weights are compressed into an optimized file format (**GGUF**).
- Once installed, Ollama exposes a **local API** on your machine (same idea as a cloud API, but local).
- More efficient and simpler than Transformers, but supports a smaller/fixed set of models.

---

## Key Term: Inference
**Inference** = running a model — i.e., giving it new input and getting an output back.
