# Context Windows & API Costs

## Context Window
- **Definition**: the maximum number of tokens a model can "look back on" at once — the max size of input it can handle.
- If input exceeds it → the model fails / errors out.

### What Actually Counts Toward the Context Window?
Not just your latest message — it includes the **entire conversation so far**:
- Every previous user message
- Every previous model reply
- The current message
- All tokens the model has generated so far in its current response

- The model generates output **one token at a time**, re-feeding the entire growing sequence back in each time.
- So the context window must fit: original input + all replies + current message + all generated tokens (up to but not including the final token).

### Why It Matters
- Governs how much the model can "remember" — references, prior context, instructions, etc.
- Important for:
  - **Multi-shot prompting**: giving several example Q&A pairs in the input for the model to learn from.
  - **RAG** and other inference-time techniques — they rely on cramming more relevant info into the input.
- Example: fitting the entire works of Shakespeare would need a context window of ~1 million tokens — only a few models (like Gemini) can handle that.

## API Costs

### Subscriptions vs API
- **Chat products** (ChatGPT, etc.): monthly subscription ($20–$200/month), usage-limited but not charged per call.
- **API usage**: pay **per call**, regardless of any subscription — this is how you'd build your own product on top of these models.
- Costs go toward paying for the actual compute (inference) — trillions of calculations per call.

### What You Pay For
- Cost depends on:
  - **Input tokens** — includes the *entire* conversation history sent in (not just your latest message).
  - **Output tokens** — includes any hidden reasoning/thinking tokens generated by reasoning models, even if you never see them.
- Both cost money because real compute happens for all of it — even reasoning you don't see.

### Example Costs (GPT-5, from Vellum leaderboard)
| | Value |
|---|---|
| Context window | 400,000 tokens |
| Input cost | $1.25 / million tokens |
| Output cost | $10 / million tokens |

- $10 could generate roughly "the complete works of Shakespeare" in output.
- For casual/small individual use (short prompts), actual costs are tiny — fractions of a cent.

### GPT-5 Nano (smallest variant)
| | Value |
|---|---|
| Input cost | $0.05 / million tokens |
| Output cost | $0.40 / million tokens |

- Extremely cheap — generating Shakespeare's complete works would cost under $1.

### Caching
- If the same input is sent repeatedly within a short time, some providers **cache** it and charge less.
- **GPT**: automatic caching.
- **Claude**: caching is available but requires more manual handling.

## Key Terms
- **Context window**: max number of tokens (input + output history) a model can process/consider at once.
- **Multi-shot prompting**: providing multiple example Q&A pairs in the input to guide the model's response.
- **Input tokens**: tokens sent *to* the model (includes full conversation history).
- **Output tokens**: tokens *generated* by the model (includes hidden reasoning tokens, if any).
- **Caching**: reusing previously processed input to reduce cost on repeated/similar calls.
