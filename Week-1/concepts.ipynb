## LLM Engineering Basics


### Closed Source Models (also called Proprietary or Frontier Models)

**Simple definition:** A closed source LLM is a model built and owned by a company that **does not share** its inner workings with the public.

**What stays hidden:**
- **Model weights** — the learned numbers that make the model work (you cannot download them)
- **Training data** — the text and data used to teach the model
- **Training methods** — the exact techniques and recipes used to build it

**How you use them:** You access these models through a website or API (like ChatGPT or Claude). You send a prompt and get a response, but you cannot run, copy, or modify the model yourself.

**Why companies do this:** It protects their investment, keeps them competitive, and lets them control how the model is used.

**Examples:** Claude (Anthropic), GPT-4 / ChatGPT (OpenAI), Gemini (Google), Grok (xAI)

**Note:** *Frontier models* means the most powerful, state-of-the-art models available. Most frontier models today are closed source.

### Open Source Models (also called Open-Weight Models)

**Simple definition:** An open source LLM is a model whose **weights are publicly released**, so anyone can download, run, study, and modify it.

**What is usually shared:**
- **Model weights** — you can download and run the model on your own computer or server
- **Architecture details** — how the model is structured
- **Usage license** — rules for how you can use and share it

**What you can do with them:**
- Run the model locally (no internet or API needed)
- Fine-tune it on your own data
- Inspect and experiment with how it works
- Build your own apps on top of it

**Important nuance:** "Open source" in AI often means **open weights**, not everything is always open. Training data and full training recipes are sometimes still kept private, even when weights are released.

**Why this matters:** Open models give you more control, privacy, and flexibility — but you need your own hardware (or cloud) to run them, and smaller open models may be less capable than top closed source ones.

**Examples:** Llama (Meta), Mixtral (Mistral), Qwen (Alibaba), Gemma (Google), GPT-OSS (OpenAI), DeepSeek (DeepSeek)

### Quick Comparison

| | Closed Source | Open Source |
|---|---|---|
| **Can you download the model?** | No | Yes |
| **Can you run it locally?** | No (API only) | Yes |
| **Can you fine-tune it?** | No (usually) | Yes |
| **Who controls updates?** | The company | You (or the community) |
| **Typical cost** | Pay per use (API) | Free to download; pay for compute |
| **Best for** | Easy access to top performance | Customization, privacy, learning |
