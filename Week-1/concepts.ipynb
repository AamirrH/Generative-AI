## LLM Engineering Basics


**Closed Source/ Frontier Models** - Models whose dataset, techniques, parameters or weights havent been revealed to for anyone to see.\
Examples - Claude(Anthropic), Gemini(Google), ChatGPT(OpenAI), Grok(xAI)

**Open Source Models** - Models whose datasets, techniques and parameters to train it have been revealed completely for anyone to tweak or tinker with.\
Examples - Llama (Meta), Mixtral (Mistral), Qwen (Alibaba), Gemma (Google), GPT-OSS (OpenAI), Deepseek (Deepseek).
